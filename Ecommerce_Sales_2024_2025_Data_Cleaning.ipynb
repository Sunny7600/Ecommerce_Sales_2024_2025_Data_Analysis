```python
print(df.groupby("Order_Status")[["Shipping_Delay_Days", "Customer_Rating"]].count())
```

```python
cancelled = df["Order_Status"] == "Cancelled"
df.loc[cancelled, ["Shipping_Delay_Days", "Customer_Rating"]] = np.nan
```

```python
print(df.groupby("Order_Status")[["Shipping_Delay_Days", "Customer_Rating"]].count())
```

```python
print(df.groupby("Region")["Shipping_Delay_Days"].mean())
```

```python
month = df["Order_Date"].dt.to_period("M")
print(df[df["Region"] == "West"].groupby(month)["Shipping_Delay_Days"].mean().tail(8))
```

```python
print(df[df["Region"] != "West"].groupby(month)["Shipping_Delay_Days"].mean().tail(8))
```

```python
print(df.groupby(["Region", df["Order_Date"] >= "2025-07-01"])[["Shipping_Delay_Days", "Customer_Rating"]].mean())
```

```python
delay_median = df.groupby(["Region", "Order_Status", month])["Shipping_Delay_Days"].transform("median")
```

```python
delay_median_region = df.groupby(["Region", "Order_Status"])["Shipping_Delay_Days"].transform("median")
```

```python
delay_median = delay_median.fillna(delay_median_region)
```

```python
fill_mask = df["Shipping_Delay_Days"].isna() & (df["Order_Status"] != "Cancelled")
df.loc[fill_mask, "Shipping_Delay_Days"] = delay_median[fill_mask]
```

```python
df[["Shipping_Delay_Days", "Customer_Rating"]] = df[
    ["Shipping_Delay_Days", "Customer_Rating"]
].round().astype("Int64")
```
